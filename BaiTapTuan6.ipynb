## W6A1
Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng giữ nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả.

Yêu cầu in ra đúng dạng list Python, ví dụ `[1, 2, 3]`

**Ví dụ:**
```
Input:  1 2 3 2 4 1 5
Output: [1, 2, 3, 4, 5]
```

````python
# Quản Đức Hiệp 0.2
my_lst = list(map(int,input().split()))
res = []
for x in my_lst:
    if x not in res:
        res.append(x)
print(res)
````

## W6A2
Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu tới vị trí i), giữ nguyên thứ tự.

**Ví dụ:**
```
Input:  1 2 3 4 5

```

````python
ds = list(map(int, input().split())) # Ta Van Nam 0.1
l = []
tong = 0
for x in ds:
    tong += x
    l.append(tong)
print(l)
    
````

## W6A3
Nhập một tuple số nguyên (một dòng) và số nguyên k (dòng kế). In ra tuple sau khi xoay trái k vị trí.

(Giữ nguyên thứ tự tương đối sau xoay; không sắp xếp.)

**Ví dụ:**
```
Input:  1 2 3 4 5
        2
Output: (3, 4, 5, 1, 2)
```

````python
lst=list(map(int,input().split())) #Truong Le Phuc Loc 0.1
k=int(input())
n=[]
if len(lst)>0:
    k=k%len(lst)
    n=lst[k:]+lst[:k]
print(n)         
````

## W6A4
Nhập một dòng gồm các cặp `key:value` cách nhau bởi dấu cách. Tạo dictionary ánh xạ mỗi key → list các value theo thứ tự xuất hiện. Nếu key lặp lại, thêm vào list của key đó.

**Ví dụ:**
```
Input:  a:1 b:2 a:3 c:4 b:5
Output: {'a': ['1', '3'], 'b': ['2', '5'], 'c': ['4']}
```

````python
#Nguyen Ha My - 0.5
lst = input().split()
d = {}
for x in lst:
    k, v = x.split(':')
    if k not in d:
        d[k] = [v]
    else:
        d[k].append(v)
print(d)
````

## W6A5
Nhập một list số nguyên. In ra dictionary `{'positives': ..., 'negatives': ..., 'zeros': ...}` (đếm số >0, <0, =0).

**Ví dụ:**
```
Input:  3 -1 0 7 -2 0 5
Output: {'positives': 3, 'negatives': 2, 'zeros': 2}
```

````python
nums = list(map(int,input().split())) # phan dinh vu
print(nums)
d = {'positives':0,'negatives':0 ,'zeros':0 }
for x in nums:
    if x>0 :
        d ['positives' ]+=1
    elif x < 0:
        d ['negatives'] += 1
    else :
        d ['zeros'] += 1
print(d)
        
````
