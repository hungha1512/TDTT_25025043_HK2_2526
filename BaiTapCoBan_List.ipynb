## Truy cập các phần tử và slicing

````python
# nums = [1, 2, 3, 4, 5]
# Lay phan tu dau tien trong nums
print(f"Phan tu dau tien: {nums[0]}")

# Lay phan tu cuoi cung trong nums
print(f"Phan tu cuoi cung: {nums[4]}")
print(f"Phan tu cuoi cung: {nums[-1]}")
print(f"Phan tu cuoi cung: {nums[len(nums)-1]}")

# Lay phan tu thu 2->4
print(f"Phan tu thu 2 den thu 4 trong list la: {nums[1:4]}")
# Lay 3 phan tu dau
print(f"3 phan tu dau tien trong list: {nums[:3]}")
# Lay tat ca phan tu o vi tri chan
print(f"Cac phan tu o vi tri chan trong list: {nums[::2]}")
````

## Thao tác trên List thường dùng

````python
nums = [1, 2, 3, 4, 5]
# Lay do dai
print(f"Do dai cua list nums: {len(nums)}")
# Them 1 phan tu vao cuoi List
print(f"List ban dau: {nums}")
nums.append(7)
print(f"List sau khi append(7) la: {nums}")
# Them 1 phan tu vao vi tri bat ky trong List
nums.insert(1, 10)
print(f"List sau khi insert(1, 10) la: {nums}")
nums.insert(6, 10)
print(f"List sau khi insert(2, 10) la: {nums}")
# Xoa 1 phan tu trong List
nums.remove(10)
print(f"List sau khi remove(10) la: {nums}")
# Xoa phan tu cuoi cung trong List
n = nums.pop()
print(f"Phan tu duoc xoa: {n}")
print(f"List sau khi pop() la: {nums}")
# Xoa phan tu thu 3 trong List -> chi so la 2
n = nums.pop(2)
print(f"Phan tu thu 3 duoc xoa la: {n}")
print(f"List sau khi pop(3) la: {nums}")

# Mo rong List bang extend
print(f"List truoc khi extend: {nums}")
nums.extend([20, 30, "hung"])
print(f"List sau khi extend: {nums}")

# Mo rong List bang cach cong 2 List
numsA = [2, 4, 6, 8, 10]
numsB = [10, 3, 5, 7, 9]
numsSum = numsA + numsB
print(f"List numsSum sau khi tong hop la: {numsSum}")
````

## Duyet List

````python
nums = [3, 5, 7]

for x in nums:
    print(f"Duyet theo gia tri: {x}")
    
for i in range(len(nums)):
    print(f"Duyet theo chi so: {nums[i]}")
    
for i, x in enumerate(nums):
    # In ra chi so dung truoc gia tri
    print(i, x)
    
for i, x in enumerate(nums, 1):
    # Tang chi so i len bat dau tu 1
    print(i, x)
````

## Hàm hay dùng của List

````python
nums = [5, 1, 9]
print(f"Do dai List: {len(nums)}")
print(f"Tong cac gia tri cua List: {sum(nums)}")
print(f"Gia tri lon nhat trong List: {max(nums)}")
print(f"Gia tri nho nhat trong List: {min(nums)}")

# Sap xep gia tri trong nums theo chieu tang dan hoac giam dan
print(f"List sau khi sap xep: {sorted(nums)}")

# Phuong thuc dao nguoc lai nums
nums.reverse()
print(f"List sau khi dao nguoc: {nums}")

# nums_reversed = nums[::-1]
# print(f"List dao nguoc: {nums_reversed}")
print(f"List hien tai: {nums}")

# Copy sau, thay doi nums_copy khong thay doi nums va nguoc lai
nums_copy = nums.copy()
nums_copy = list(nums)
# Copy nong, thay doi nums_copy lam thay doi den nums
nums_copy = nums
print(f"List copy: {nums_copy}")
````

## Duyet dict

````python
student = {"name": "Hung", "age": 23}

# Duyet bang key
for key in student:
    print(key, student[key])
    
# Duyet bang item
for key, value in student.items():
    print(key, value)
    
for key in student.keys():
    print(key)
    
for value in student.values():
    print(value)
````

### Bài 2: Tìm số nhỏ nhất của một danh sách các số nguyên. KHONG DUNG HAM SAN CO CUA LIST

````python
# lst = list(map(int, input().split()))

def min_of_list (lst):
    if not lst:
        return None
    m = lst[0]
    for x in lst:
        if x < m:
            m = x
    return m

assert min_of_list([]) == None
assert min_of_list([23, 67, 900, -10]) == -10, "Sai test nay roi ban oi"
assert min_of_list([575, 939, 697, 138, 183, 386, 147, 695, 622, 289]) == 138

print("Pass all Tests!!!!!!!!")
````

### Bài 2b: Tìm số nhỏ nhất của 1 list và vị trí của nó, nếu có nhiều số nhỏ nhất bằng nhau thì lấy chỉ số đầu tiên (nhỏ nhất)
10 0 0 12 34 5 6 7  8 89 0 1
0 1

````python


def min_and_pos(lst):
    if not lst:
        return None
    min_val = lst[0]
    pos_val = 0
    for i, x in enumerate(lst):
        if x < min_val:
            min_val = x
            pos_val = i
            
    return min_val, pos_val + 1 

assert min_and_pos([10, 0, 0, 12, 34, 5, 6, 7, 8, 89, 0, 1]) == (0, 2)
assert min_and_pos([1, 2, 3, 4,6,7]) == (1, 1)
assert min_and_pos([]) == None

print("Pass all Tests!!!")
````

### Bài 3: 
Tìm giá trị trung bình của các số nguyên dương trong một list.
Nếu không có số dương nào thì trả về "invalid"

````python
#Tuan
def Average_positive_numbers_of_list(lst):
    if not lst:
        return 'invalid'
    s = 0
    cnt = 0
    for x in lst:
        if x > 0:
            s += x
            cnt += 1
    if cnt == 0:
        return "invalid"
    return round(s/cnt, 2)
    
assert Average_positive_numbers_of_list([1, 2, 6, 1, -10, -10]) == 2.50
assert Average_positive_numbers_of_list([-2, -1, -20, -343]) == "invalid"
assert Average_positive_numbers_of_list([]) == "invalid"

print("Accepted!")
````
