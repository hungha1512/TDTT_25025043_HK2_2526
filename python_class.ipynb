
## 4. Class variable (Biến của lớp) và Instance variable(Biến của đối tượng)

2. Class variable (Biến của lớp)
👉 Là biến dùng CHUNG cho tất cả các object thuộc lớp
   - Khai báo trực tiếp trong class

````python

class Student:
    school = "PTIT"   # class variable

    def __init__(self, name):
        self.name = name  # instance variable

s1 = Student("A")
s2 = Student("B")

print(s1.school, s2.school)
Student.school = "VNU" # Thay đổi giá trị của class variable

print(s1.school, s2.school)

print(s1.name, s2.name)
s1.name = "C"
print(s1.name, s2.name)
````

````python
# Merge sort

def merge(a, b):
    res = []
    i, j = 0, 0
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            res.append(a[i])
            i += 1
        else: 
            res.append(b[j])
            j += 1
    
    res.extend(a[i:])
    res.extend(b[j:])
    
    return res
    
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    return merge(left, right)

print(merge_sort([2, 5, 2, 3, 7, 4]))
````


## 5. Các phương thức đặc biệt (__dunder__ methods)
#### Đọc thêm bên file kia :D 


````python

class Point:
    def __init__(self, _x, _y): # hàm - phương thức khởi tạo
        self.x = _x
        self.y = _y

    def __str__(self): #hàm - phương thức in (print) đối tượng
        return f"({self.x}, {self.y})"

    def __add__(self, other):# Hàm - phương thức cộng đối tượng với 1 đối tượng khác
        return Point(self.x + other.x, self.y + other.y)
    
    def __del__ (self):#Hàm - Phương thức hủy đối tượng - Tự động được gọi khi kết thúc
        print("Object deleted")

p1 = Point(1, 2)
p2 = Point(3, 4)
print(p1)
print(p1 + p2)
````


## 6. Kế thừa (Inheritance)


Kế thừa là cho phép lớp con (subclass) sử dụng lại thuộc tính (attribute) và phương thức (method) của lớp cha (superclass), đồng thời có thể mở rộng và ghi đè.

Thuật ngữ:
- Lớp cha = superclass = base class = parent class
- Lớp con = subclass = dervied class = child class

````python
class <tên lớp con>(<tên lớp cha>):
    <nội dung>
````

````python
class Animal:
    def __init__(self, name):
        self.name = name
        
    def speak(self):
        return "..."
    
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

    def speak(self): # Overriding
        return "Woo Woo"

class Human(Animal):
    pass # Không có __init__ => Tự động dùng Animal.__init__

dog = Dog("Lucky", "Chihuahua")
print(dog.name, dog.breed)
print(dog.speak())

human = Human("A")
print(human.name)
````

### Quy tắc tìm phương thức từ dưới lên:

Khi gọi ```obj.method()```, Python tìm ```method()``` theo thứ tự:
1. Tìm class của chính ```obj```
2. Nếu không có, tìm lên lớp cha
3. Tiếp tục tìm các lớp trên cho đến khi tìm thấy
4. Nếu không tìm thấy => ```AttributeError```

````python
class Account:
    def __init__(self, balance = 0):
        self._balance = balance
    
    def deposit(self, amount):
        self._balance += amount

class InteresetAccount(Account):
    def addInterest(self, rate):
        self.deposit(self._balance * rate)
        
acct = InteresetAccount(1000)
acct.deposit(500)
acct.addInterest(0.05)

print(acct._balance)

print(isinstance(acct, Account))
print(issubclass(Account, InteresetAccount))
````

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y 
        
    def __eq__(self, value):
        return type(value) == Point and self.x == value.x and self.y == value.y
        
p1 = Point(0, 0)
p2 = Point(0, 0)

print(p1 == p2)
````

# Ngoai le (Exception)

1. Cấu trúc try-except-else-finally:

```
try:
    <code có thể gây lỗi>
except <Loại lỗi>:
    <xử lý khi có lỗi>
else:
    <chạy khi không có lỗi>
finally:
    <luôn luôn chạy kể cả có lỗi hay không>
```

````python
try:
    x = int(input())
    print(x)
    print(10/x)
except ValueError:
    print("Khong phai so nguyen")
except ZeroDivisionError:
    print("Khong chia cho so 0")
else:
    print("Tinh thanh cong")
finally: 
    print("Ket thuc")
````

````python
try:
    lst = [1, 2, 3]
    print(lst[10])
except (IndexError, KeyError) as e:
    print(f"Loi: {e}")
````

Raise: tu nem loi khi dau vao khong hop le

````python
def max_and_min(nums):
    if len(nums) == 0:
        raise ValueError("danh sach bi trong")
    return max(nums), min(nums)

try:
    max_and_min([])
except ValueError as e:
    print(e)
````

Tự viết ngoại lệ

````python
class NegativeValuesError(Exception):
    def __init__(self, value):
        super().__init__(f"Gia tri khong duoc am: {value}")
        self.value = value

def set_age(age):
    if age < 0:
        raise NegativeValuesError(age)
    
    return age

try:
    set_age(-5)
except NegativeValuesError as e:
    print(e)
````

````python

class Account:
    def __init__(self, balance):
        self._balance = balance

    def get_balance(self):
        return self._balance

acc = Account(1000)
print(acc.get_balance())
````


## 9. Property (@property)


````python

class Circle:
    def __init__(self, r):
        self._r = r

    @property
    def area(self):
        return 3.14 * self._r ** 2

c = Circle(5)
print(c.area)
````


## 10. Classmethod & Staticmethod


````python

class MathUtils:
    pi = 3.14

    @classmethod
    def circle_area(cls, r):
        return cls.pi * r * r

    @staticmethod
    def add(a, b):
        return a + b

print(MathUtils.circle_area(3))
print(MathUtils.add(2, 5))
````


## 12. Ví dụ tổng hợp (Mini Project)


````python

class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Không đủ tiền")

    def __str__(self):
        return f"{self.owner}: {self.balance}"

acc = BankAccount("Long", 1000)
acc.deposit(500)
acc.withdraw(300)
print(acc)
````
