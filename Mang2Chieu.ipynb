# Mảng 2 chiều trong Python

Mảng 2 chiều là một mảng chứa nhiều mảng, hay gọi là 1 list chứa một hoặc nhiều list con.

Phổ biến nhất là biểu diễn ma trận, dữ liệu (hàng cột).

## Khởi tạo và truy cập mảng 2 chiều

````python
# Ma trận 3x3
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]   
]

matrix_2 = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

# Lấy phần tử có giá trị 2 trong list matrix
print(matrix[0][1])
# Lấy phần tử có giá trị 7 trong list matrix_2
print(matrix_2[2][0])

matrix_3 = [[1, 2, 3, [4, 5, 6, [7, 8, 9]]]]
# Lấy phần tử có giá trị 8 trong list matrix_3
print(matrix_3[0][3][3][1])
````

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]   
]

# Duyệt theo chỉ số
rows = len(matrix)
cols = len(matrix[0])
for i in range(rows):
    for j in range(cols):
        print(f"Phần tử tại [{i}][{j}] của matrix là {matrix[i][j]}")
        
print("----------------------")

# Duyệt theo chỉ số mảng có mảng con không đồng nhất độ dài
matrix_2 = [
    [1, 2, 3, 11, 12],
    [4, 5, 6, 10],
    [7, 8, 9]   
]

for i in range(len(matrix_2)):
    for j in range(len(matrix_2[i])):
        print(f"Phần tử tại [{i}][{j}] của matrix_2 là {matrix_2[i][j]}")
        
````

## Thao tác hay gặp

````python
# Làm phẳng mảng/danh sách

matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
# result = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Làm phẳng bằng append
result = []
for row in matrix:
    for item in row:
        result.append(item)
        
print(result)

# Làm phẳng bằng extend
result_2 = []
for row in matrix:
    result_2.extend(row)
print(result_2)

# Làm phẳng bằng list comprehension
result_3 = [item for row in matrix for item in row]
print(result_3)
````

````python
# Tính tổng theo hàng 
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]   
]

count = 1
for row in matrix:
    sum = 0
    for item in row:
        sum += item
    print(f"Tong cua hang {count} la: {sum}")
    count += 1

def tong_hang(matrix, k):
    sum = 0
    for i in range(len(matrix[k-1])):
        sum += matrix[k-1][i]
    return sum

print(tong_hang(matrix, 3))

# Tính tổng cột
def tong_cot(matrix, k):
    sum_column = 0
    if k < 0 or k >= len(matrix):
        return None
    for i in range(len(matrix)):
        sum_column += matrix[i][k]
    return sum_column

print(tong_cot(matrix, 2))
    
````
